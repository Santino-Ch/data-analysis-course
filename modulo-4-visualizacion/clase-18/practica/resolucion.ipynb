```python
"""PREPARATIVOS"""
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("sensores.csv") 
print(df.head())
```

```python
"""EJERCICIO 1"""
df["dias"] = df.index % 5
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(df.index, df["produccion"])
ax.set_title("Producción a lo largo del tiempo")
ax.set_xlabel("Observación en dias") #eje x
ax.set_ylabel("Producción") #eje y
ax.set_ylim(0, df["produccion"].max()) #para no exagerar las proporciones
ax.grid(True, alpha=0.3)


plt.show()
```

```python
"""EJERCICIO 2"""
df["dias"] = df.index % 24
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(df.index, df["temperatura"])
ax.set_title("Temperatura a lo largo del tiempo")
ax.set_xlabel("Observación en horas") #eje x
ax.set_ylabel("Temperatura") #eje y
ax.set_ylim(0, df["temperatura"].max()) #para no exagerar las proporciones
ax.grid(True, alpha=0.3)


plt.show()

```

```python
"""EJERCICIO 3"""
consumo_por_turno = df.groupby("turno", observed=True)["consumo_kwh"].mean().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(7, 5))
ax.bar(consumo_por_turno.index, consumo_por_turno.values, color="#41ab5d")
ax.set_title("Consumo eléctrico promedio por turno")
ax.set_ylabel("Consumo (kWh)")
plt.show()
```
